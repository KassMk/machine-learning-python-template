```python
df_clean = df.copy()

for col in df_clean.columns:
    if "Rainfall" in col:
        df_clean[col] = df_clean[col].fillna(0)

for col in df_clean.columns:
    if "Groundwater" in col or "Volume" in col or "Hydrometry" in col:
        df_clean[col] = df_clean[col].interpolate(method="time")

for col in df_clean.columns:
    if "Temperature" in col:
        df_clean[col] = df_clean[col].fillna(method="ffill")

print("Nulos después de limpieza:\n", df_clean.isna().sum().sort_values(ascending=False).head())
```

```python
target_col = "Depth_to_Groundwater_LT2"

df_clean[target_col] = df_clean[target_col].interpolate(method="linear")

df_clean[target_col] = df_clean[target_col].fillna(method="ffill").fillna(method="bfill")

print("Nulos restantes en target:", df_clean[target_col].isna().sum())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
df_clean["year"] = df_clean.index.year
df_clean["month"] = df_clean.index.month
df_clean["dayofyear"] = df_clean.index.dayofyear
df_clean["week"] = df_clean.index.isocalendar().week.astype(int)
df_clean["quarter"] = df_clean.index.quarter

# --- Lags ---
target_col = "Depth_to_Groundwater_LT2"
for lag in [1, 7, 30]:
    df_clean[f"{target_col}_lag{lag}"] = df_clean[target_col].shift(lag)

# --- Rolling means ---
df_clean[f"{target_col}_roll7"] = df_clean[target_col].rolling(7).mean()
df_clean[f"{target_col}_roll30"] = df_clean[target_col].rolling(30).mean()

# --- Eliminar NaNs generados ---
df_model = df_clean.dropna().copy()

print("Shape final del dataset:", df_model.shape)
```

```python
X = df_model.drop(columns=[target_col])
y = df_model[target_col]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False
)

print("Train size:", X_train.shape, "Test size:", X_test.shape)
```

```python
rf = RandomForestRegressor(
    n_estimators=200, random_state=42, n_jobs=-1
)

rf.fit(X_train, y_train)
y_pred = rf.predict(X_test)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("RMSE:", rmse)
print("R2:", r2)
```

```python
plt.figure(figsize=(12,6))
plt.plot(y_test.index, y_test, label="Real", color="blue")
plt.plot(y_test.index, y_pred, label="Predicho (RF)", color="red", alpha=0.7)
plt.title(f"RandomForest Predicción de {target_col}")
plt.xlabel("Fecha")
plt.ylabel("Profundidad (m)")
plt.legend()
plt.show()
```

```python
mean, std = df_clean[target_col].mean(), df_clean[target_col].std()
mask = np.abs(df_clean[target_col] - mean) > 3*std
print("Outliers detectados en target:", mask.sum())

df_clean.loc[mask, target_col] = np.nan
df_clean[target_col] = df_clean[target_col].interpolate(method="linear").fillna(method="bfill").fillna(method="ffill")

df_clean_model = df_clean.dropna().copy()

X = df_clean_model.drop(columns=[target_col])
y = df_clean_model[target_col]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False
)

rf = RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)
y_pred = rf.predict(X_test)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)
print("RandomForest después de limpiar outliers → RMSE:", rmse, " R2:", r2)
```

```python
ts = df_clean[target_col]

train_size = int(len(ts) * 0.8)
train, test = ts[:train_size], ts[train_size:]

model = ARIMA(train, order=(5,1,0))
model_fit = model.fit()

print(model_fit.summary())

pred = model_fit.forecast(steps=len(test))

rmse_arima = np.sqrt(mean_squared_error(test, pred))
r2_arima = r2_score(test, pred)
print("ARIMA → RMSE:", rmse_arima, " R2:", r2_arima)

plt.figure(figsize=(12,6))
plt.plot(test.index, test, label="Real", color="blue")
plt.plot(test.index, pred, label="Predicho (ARIMA)", color="green")
plt.title("ARIMA Predicción Depth_to_Groundwater_LT2")
plt.xlabel("Fecha")
plt.ylabel("Profundidad (m)")
plt.legend()
plt.show()
```

```python
df_features = df_clean.copy()

for lag in [1, 7, 30, 90, 180]:
    df_features[f"{target_col}_lag{lag}"] = df_features[target_col].shift(lag)

for window in [7, 30, 90, 180]:
    df_features[f"{target_col}_roll{window}"] = df_features[target_col].rolling(window).mean()

df_features["sin_day"] = np.sin(2 * np.pi * df_features.index.dayofyear / 365)
df_features["cos_day"] = np.cos(2 * np.pi * df_features.index.dayofyear / 365)

df_features["month"] = df_features.index.month
df_features["year"] = df_features.index.year

df_features = df_features.dropna()
```

```python
X = df_features.drop(columns=[target_col])
y = df_features[target_col]

train_size = int(len(X) * 0.8)
X_train, X_test = X.iloc[:train_size], X.iloc[train_size:]
y_train, y_test = y.iloc[:train_size], y.iloc[train_size:]
```

```python
rf = RandomForestRegressor(
    n_estimators=300, max_depth=15, random_state=42, n_jobs=-1
)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)

xgb = XGBRegressor(
    n_estimators=500, learning_rate=0.05, max_depth=6,
    subsample=0.9, colsample_bytree=0.9, random_state=42, n_jobs=-1
)
xgb.fit(X_train, y_train)
y_pred_xgb = xgb.predict(X_test)

rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)

rmse_xgb = np.sqrt(mean_squared_error(y_test, y_pred_xgb))
r2_xgb = r2_score(y_test, y_pred_xgb)

print("RandomForest → RMSE:", rmse_rf, " R²:", r2_rf)
print("XGBoost → RMSE:", rmse_xgb, " R²:", r2_xgb)
```
